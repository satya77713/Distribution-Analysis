```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/Employee.csv')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python

df.info()

print("----------------------------------------------------------------")
display(df.describe())
```

```python
age = df['Age'].dropna()

print("Selected Column: Age")
print("Total Records:", len(age))
print("Minimum Age:", age.min())
print("Maximum Age:", age.max())
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    age,
    bins=10,
    edgecolor='black'
)

plt.title('Distribution of Employee Age')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.hist(age, bins=15, edgecolor='black')

plt.title('Employee Age Distribution (15 Bins)')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

```python
print("Mean Age:", age.mean())
print("Median Age:", age.median())
print("Mode Age:", age.mode().tolist())
print("Standard Deviation:", age.std())
print("Skewness:", age.skew())
```

```python
skewness = age.skew()

if skewness > 0.5:
    print("Distribution is Positively Skewed (Right Skewed)")

elif skewness < -0.5:
    print("Distribution is Negatively Skewed (Left Skewed)")

else:
    print("Distribution is Approximately Symmetrical")
```

```python
print("OBSERVATIONS FROM HISTOGRAM")

print("1. The histogram shows the frequency distribution of employee ages.")

print("2. Mean and median help understand the central tendency of employee ages.")

print("3. Skewness indicates whether the age distribution is symmetrical or skewed.")

print("\nSkewness Value:", round(skewness, 3))
```

```python
plt.figure(figsize=(10, 6))

plt.hist(age, bins=10, edgecolor='black')

plt.axvline(
    age.mean(),
    linestyle='--',
    label='Mean Age'
)

plt.axvline(
    age.median(),
    linestyle=':',
    label='Median Age'
)

plt.title('Employee Age Distribution with Mean and Median')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
